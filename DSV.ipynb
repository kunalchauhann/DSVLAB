```python
import pandas as pd
data = {
    "Name":["Kunal","Rahul","Atul"],
    "Age":[21,23,25],
    "City":["Farrukhabad","Mathura","Agra"]
}
df=pd.DataFrame(data)
print(df)
```

```python
print(student_df["City"])
```

```python
data = {
    "Book":["Java","C++","Python"],
    "Author":["Kunal","Atul","Kartik"],
    "Price":[500,400,600]
}
book_df=pd.DataFrame(data)
print(book_df)
```

```python
book_df["Publisher"] = "Penguin"
print(book_df)
```

```python
import numpy as np

df = pd.DataFrame(index=range(5))

df["SerialNo"] = np.arange(5)

print(df)
```

```python
data = {
    "Movie":["Vikings","GOT","Last of us"],
    "Year":[2011,2015,2014],
    "Rating":["****","*****","*****"]
}
movie_df=pd.DataFrame(data)
print(movie_df)
```

```python
new_df = movie_df[["Movie", "Rating"]]

new_df["Director"] = ["Director1", "Director2", "Director3"]

print(new_df)
```

```python
city_series = pd.Series(["Mathura", "Delhi"],index=[0, 3])
print(city_series)
```

```python
branch = pd.Series(["Mathura", "Delhi"],index=[0, 3])
df = pd.DataFrame(index=range(4))
df["Branch"] = branch

print(df)
```

```python
data = {
    "Employee":["Kunal","Rahul","Atul"],
    "Department":["Tech","Marketing","Sales"],
    "Experience":["2Year","3Year","Freasher"]
}
employee_df=pd.DataFrame(data)
print(employee_df)
```

```python
print(employee_df["Department"])
print(employee_df.columns)
```

```python
employee_df["Office"] = "Head Office"
print(employee_df)
```

```python
employee_df["Office"] = np.arange(3)
print(employee_df)
```

```python
office_names = pd.Series(
    ["Delhi", "Mumbai", "Pune"],
    index=[0, 2, 4]
)
employee_df["Office"] = office_names
print(employee_df)
```
